```python
import numpy as np    # Linearly equally spaced number
arr1 = np.linspace(1,100,18).reshape(2,3,3)   # reshape(Matrices, Row, Column)
print(arr1)
```

```python
# Slicing
arr_slice = arr1[1,:,:]
arr_slice
```

```python
print(arr1[1] - arr1[0])
```

```python
# flattening
# It reduces dimension to one dimention
# Flatten and ravel are same output but mostly use flatten here
print(arr_slice)
print("Dimension of Original array is: ",arr_slice.ndim)
print("----------------------------------------------------")
print(arr_slice.flatten())   # Copy
print("Dimension of array is: ",arr_slice.flatten().ndim)
print("----------------------------------------------------")
print(arr_slice.ravel())     # view only and we can change
print("Dimension of array is: ",arr_slice.ravel().ndim)
print("----------------------------------------------------")
```
