# Web Scrapping

```python
import requests
from bs4 import BeautifulSoup
```

```python
# 
url = "https://www.flipkart.com/search?q=women+hand+bag&sid=reh%2Cihu&as=on&as-show=on&otracker=AS_QueryStore_OrganicAutoSuggest_1_15_na_na_na&otracker1=AS_QueryStore_OrganicAutoSuggest_1_15_na_na_na&as-pos=1&as-type=HISTORY&suggestionId=women+hand+bag%7CHandbags+%26+Clutches&requestId=1f2e7770-b324-42d4-9dd6-afd0a34cb8be"

headers = {
    "User-Agent":(
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/115.0.0.0 Safari/537.36"
            ),
    "Referer": "https://www.flipkart.com"
    }
response = requests.get(url, headers=headers)
print(response)    # 200 : Means data can be Scrapped from that Webpage.
```

```python
Soup = BeautifulSoup(response.content ,"html.parser")
```

```python
names_list = Soup.find_all("div",class_="Fo1I0b")
names_list
```

```python
LengthOfFridge = len(Soup.find_all("div",class_="Fo1I0b"))
LengthOfFridge
```

```python
list1 = [7,3,8,4,9,5,6,1,100,102,103,104]
print("Length:", len(list1))
# One Dimension
arr1 = np.array(list1)
print("One Dimension:")
print(arr1)
print("Dimensions:", arr1.ndim)
print("---------------------")
# Convert it into an array of 4 row and two column
# Two Dimension (4 rows, 3 columns)
arr2 = np.array(list1).reshape(4,3)
print("Two Dimension:")
print(arr2)
print("Dimensions:", arr2.ndim)

print("---------------------")

# Three Dimension (2, 3, 2)
arr3 = np.array(list1).reshape(2,3,2)
print("Three Dimension:")
print(arr3)
print("Dimensions:", arr3.ndim)
```
