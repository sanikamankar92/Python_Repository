```python
# Program to reverse a user-entered number
num = int(input("Enter a number: "))
rev = 0
while num > 0: 
    digit = num % 10 
    rev = rev * 10 + digit 
    num = num // 10
print("Reverse number:", rev)
```

```python
# Program to check whether a number is Palindrome or not
# A palindrome number reads the same forward and backward. Example: 121,  madam is a word palindrome.
num = int(input("Enter a number: "))
temp = num
rev = 0
while num > 0: 
    digit = num % 10  
    rev = rev * 10 + digit  
    num = num // 10
if temp == rev: 
    print("Palindrome number")
else: 
    print("Not a palindrome number")
```

```python
# Program to check whether a number is Armstrong or not
# An Armstrong number is equal to the sum of its digits, each raised to the power of the number of digits. Example: \(153 = 1^3 + 5^3 + 3^3 = 153\).
num = int(input("Enter a number: "))
temp = num
total = 0
power = len(str(num))
while num > 0: 
    digit = num % 10   
    total = total + digit ** power  
    num = num // 10
if temp == total:  
    print("Armstrong number")
else:   
    print("Not an Armstrong number")
```

```python
# Program to check whether a number is Prime or not
# A prime number has exactly two factors: 1 and itself.
num = int(input("Enter a number: "))
count = 0
if num > 1:  
    for i in range(1, num + 1): 
            if num % i == 0:  
                        count = count + 1
if count == 2: 
        print("Prime number")
else:  
        print("Not a prime number")
```

```python
# Program to check whether a number is Perfect or not
# A perfect number equals the sum of its positive factors excluding itself. Example: \(6 = 1 + 2 + 3\).
num = int(input("Enter a number: "))
total = 0
for i in range(1, num): 
    if num % i == 0:   
        total = total + i
if num > 0 and total == num:
        print("Perfect number")
else:  
        print("Not a perfect number")
```
